````python
from pathlib import Path
import runpy

patch=Path("fhir_patch.py")
if not patch.exists():
    raise FileNotFoundError("Pon fhir_patch.py en la misma carpeta del proyecto.")
runpy.run_path(str(patch))
````

````python
import py_compile
py_compile.compile("main.py",doraise=True)
print("Sintaxis de main.py correcta después de agregar FHIR.")
````
